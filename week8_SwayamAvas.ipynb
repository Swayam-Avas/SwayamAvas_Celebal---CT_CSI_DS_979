```python
import json

# -----------------------------
# Calculator Tool
# -----------------------------
def calculator(expression):
    try:
        result = eval(expression)
        return str(result)
    except:
        return "Error in calculation"


# -----------------------------
# Keyword Extraction Tool
# -----------------------------
def extract_keywords(text):
    try:
        words = text.split()

        keywords = []

        for word in words:
            word = word.lower().strip(".,!?")

            if len(word) > 4 and word not in keywords:
                keywords.append(word)

        return keywords[:5]

    except:
        return []


# -----------------------------
# Smart Agent
# -----------------------------
def agent(query):

    query_lower = query.lower()

    try:

        # Check for calculation request
        if "calculate" in query_lower:

            expression = query.replace("Calculate", "").replace("calculate", "").strip()

            answer = calculator(expression)

            return {
                "type": "calculation",
                "result": answer
            }

        # Check for keyword extraction
        elif "keywords" in query_lower:

            if "from" in query_lower:
                text = query.split("from", 1)[1].strip()
            else:
                text = query

            answer = extract_keywords(text)

            return {
                "type": "keywords",
                "result": answer
            }

        # Handle general questions
        else:

            return {
                "type": "general",
                "result": "This looks like a general question. No tool was required."
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }


# -----------------------------
# Test Cases
# -----------------------------
queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query :", q)
    print("Response :", agent(q))
    print("-" * 50)


# -----------------------------
# Interactive Mode
# -----------------------------
while True:

    user_input = input("\nEnter your query (type 'exit' to quit): ")

    if user_input.lower() == "exit":
        print("Thank you! Program Closed.")
        break

    response = agent(user_input)

    print(json.dumps(response, indent=4))
```
